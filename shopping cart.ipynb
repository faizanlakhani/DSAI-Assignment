```python
products = {
    "P1": {"name": "Shoes",      "price": 2500, "stock": 5},
    "P2": {"name": "Shirt",      "price": 1500, "stock": 10},
    "P3": {"name": "Bag",        "price": 3000, "stock": 4},
    "P4": {"name": "Watch",      "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6},
}


cart = {}

while True:
    print("\n ONLINE SHOPPING CART ")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")
    choice = input("Enter your choice (1-5): ").strip()

    
    if choice == "1":
        print("\n--- Available Products ---")
        print(f"{'ID':<8}{'Product':<15}{'Price':<12}{'Stock'}")
        print("-" * 42)
        for pid, info in products.items():
            price = f"Rs. {info['price']:,}"
            print(f"{pid:<8}{info['name']:<15}{price:<12}{info['stock']}")

   
    elif choice == "2":
        pid = input("Enter product ID (e.g. P1): ").strip().upper()
        if pid not in products:
            print("Product not found. Please enter a valid product ID.")
        else:
            qty_text = input("Enter quantity: ").strip()
            if not qty_text.isdigit() or int(qty_text) <= 0:
                print("Invalid quantity. Please enter a whole number greater than 0.")
            else:
                qty = int(qty_text)
                already_in_cart = cart.get(pid, 0)
                total_requested = already_in_cart + qty
                stock = products[pid]["stock"]

                if total_requested > stock:
                    print(f"Not enough stock! Only {stock} {products[pid]['name']} available "
                          f"({already_in_cart} already in your cart).")
                else:
                    cart[pid] = total_requested
                    print(f"Added {qty} x {products[pid]['name']} to your cart.")

    
    elif choice == "3":
        if len(cart) == 0:
            print("Your cart is empty.")
        else:
            print("\n--- Your Cart ---")
            print(f"{'Product':<15}{'Qty':<6}{'Price':<12}{'Subtotal'}")
            print("-" * 48)
            cart_total = 0
            for pid, qty in cart.items():
                name = products[pid]["name"]
                price = products[pid]["price"]
                subtotal = price * qty
                cart_total += subtotal
                print(f"{name:<15}{qty:<6}{'Rs. ' + format(price, ','):<12}Rs. {subtotal:,}")
            print("-" * 48)
            print(f"Cart Total: Rs. {cart_total:,}")

    
    elif choice == "4":
        if len(cart) == 0:
            print("Your cart is empty. Add some items before checkout.")
        else:
            cart_total = 0
            print("\n--- FINAL BILL ---")
            for pid, qty in cart.items():
                subtotal = products[pid]["price"] * qty
                cart_total += subtotal
                print(f"{products[pid]['name']:<15}{qty} x Rs. {products[pid]['price']:,} = Rs. {subtotal:,}")

            if cart_total >= 5000:
                discount = cart_total * 0.10
                print("10% discount applied (cart total is Rs. 5,000 or more).")
            else:
                discount = 0
                print("No discount (cart total is below Rs. 5,000).")

            final_bill = cart_total - discount
            print("-" * 40)
            print(f"Cart Total  : Rs. {cart_total:,.2f}")
            print(f"Discount    : Rs. {discount:,.2f}")
            print(f"Final Bill  : Rs. {final_bill:,.2f}")

           
            for pid, qty in cart.items():
                products[pid]["stock"] -= qty
            cart.clear()
            print("\nCheckout complete. Thank you for shopping with us!")

   
    elif choice == "5":
        print("Thank you for visiting. Goodbye!")
        break

    else:
        print("Invalid choice. Please enter a number from 1 to 5.")
```
